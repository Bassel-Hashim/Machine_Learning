```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
dict1= {
    "Area": [90, 100, 120, 150, 200],
    "Price":[800000, 1000000, 1200000, 2000000, 3000000]
}
df =pd.DataFrame(dict1)
```

```python
df.head(

)
```

```python
#Define the features (independent variable) and target (dependent variable)
X = df[["Area"]]
Y = df[["Price"]]
```

```python
# create and train the model
model = LinearRegression()
model.fit(X, Y)
```

```python
prediction = model.predict(X)
```

```python
prediction
```

```python
#Plot the results
plt.scatter(X,Y, color='blue', label='Actual Data')
plt.plot(X,Y, color='blue', label='Actual Data')
plt. plot(X, prediction, color='red', label='Regression Line')
plt.xlabel('Area')
plt.ylabel('Price')
plt.title('Linear Regression: Area vs Price')
plt.legend()
plt.show()
```

```python
#Test the model

model.predict([[70]])
```
