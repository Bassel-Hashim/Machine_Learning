```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
```

```python
df = pd.read_csv("houses_clean.csv")
print(df.head())
print(df.shape)
```

```python
x = df.drop(columns=["price"])
y = df["price"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model= LinearRegression()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
```

```python
print("\nTrain R2 Score:", model.score (x_train, y_train))# Determines the acuuracy for train
print("Test R2 Score:", model.score (x_test, y_test))# Determines the acuuracy for test
print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE :", mean_squared_error(y_test, y_pred)** 0.5)
print("MSE :", mean_absolute_error(y_test, y_pred))
```

```python
print("\nIntercept:", model.intercept_)
for name, coef in zip(x.columns, model.coef_):
    print(f"{name}:", {round(coef,2)})
```
